### 1.While loop & Control Statements :
#### (else, break, continue, pass)
A while loop executes a block of code as long as the condition is True.

### Control Statements
- break: Stops the loop.
- continue: Skips the current iteration.
- else: Executes when the loop finishes without break.
- pass: The pass statement does nothing. It is used as a placeholder when a statement is required syntactically but no action is needed.

### Number Guessing Game
##### Problem Statement: Create a Python program that implements a simple number
##### guessing game using a while loop. The program should make use of control
##### statements such as else, break, and continue.

```python
import random
# Generate random number between 1 and 10
secret_number = random.randint(1,10)
attempts = 3

while attempts > 0:
    guess_num = int(input("Guess any number from 1 to 10 : "))
    if guess_num < 1 or guess_num > 10:
        print(" Your guess is out of range. Guess a number between 1 and 10 " )
        continue
    attempts = attempts - 1
    
    if guess_num == secret_number:
        pass
        print(" Congratulations! You guessed the correct number. ")
        break
    elif guess_num < secret_number:
        print(" Try a number above your guess ")
    else:
        print( " Try a number below your guess ")
else:
    print("Better luck next time !! ")
        
```

### 2. For Loop
#### (for, range)

A for loop executes a block of code for each item in a sequence.

### Key Concepts
- for: Repeats a block of code for each item.
- range(): Generates a sequence of numbers.
- i: A loop variable that takes each value in the sequence.

### Multiplication Table Generator
##### Problem Statement: Create a Python program that generates and prints a
##### multiplication table (from 1 to 10) for a given number using a for loop and the range
##### function.

```python

number = int(input("Enter the number for which you want the multiplication table: "))

for i in range(1, 11):
    result = number * i
    print(number, "x", i, "=", result)
```

### 3. Function:
#### (def, parameters, return)

A function is a reusable block of code that performs a specific task.

### Key Concepts
- def: Used to define a function.
- Parameters: Variables that receive values passed to a function.
- return: Sends the result back to the calling code.

### BMI Calculator

```python
def calculate_bmi(weight, height):
    result_bmi = weight / (height ** 2)
    return result_bmi

weight = float(input("Enter the weight in KG : "))
height = float(input("Enter the height in meters : "))

bmi = calculate_bmi(weight, height)

print(f"Your weight is {weight} kg")
print(f"Your height is {height} meters")

print(f"Your BMI is: {bmi}")
print(f"Your BMI is: {bmi:.2f}") # Display BMI rounded to 2 decimal places
```
